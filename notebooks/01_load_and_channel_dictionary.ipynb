# 01 - Synthetic Channel Data

This notebook loads a generated example workbook with anonymous channel identifiers. The values are synthetic and do not come from an organization or operational system.

```python
import sys
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().resolve()
while not (ROOT / "src" / "channel_io.py").is_file() and ROOT.parent != ROOT:
    ROOT = ROOT.parent
if not (ROOT / "src" / "channel_io.py").is_file():
    raise FileNotFoundError("Could not find the project root containing src/channel_io.py.")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.channel_io import find_raw_file, load_workbook, write_markdown_dictionary

RAW_FILE = find_raw_file(ROOT / "data" / "raw")
print("Using synthetic example:", RAW_FILE.name)
```

## Load and validate

The channel dictionary contains generic identifiers only. It intentionally excludes source-system names, physical units, and raw measurement ranges.

```python
values, channel_dict = load_workbook(RAW_FILE)
print("Values shape:", values.shape)
print("Channel count:", channel_dict["name"].nunique())
print("Timestamps:", values.index.min().date(), "to", values.index.max().date())

interim = ROOT / "data" / "interim"
interim.mkdir(parents=True, exist_ok=True)
values.to_csv(interim / "channel_values.csv")
channel_dict.to_csv(interim / "channel_dictionary.csv", index=False)
write_markdown_dictionary(channel_dict, ROOT / "docs" / "data_dictionary.md")
print("Saved synthetic values and generic channel metadata.")
```
